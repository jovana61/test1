This lesson covers the basics of creating classes and objects, including instance variables and methods.

```python
### Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):   #init je constructor
        self.name=name    # ovako se definisu atributi, self obavezno, mozr i druga varoijabila
        self.age=age

## create objects
dog1=Dog("Buddy",3)
print(dog1)  #<__main__.Dog object at 0x0000015A0B7E99D0>
print(dog1.name)
print(dog1.age)
    
    
```

```python
## Define a class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name=name   #  ovo je properties
        self.age=age
    
    def bark(self):     # ovo je instance method 
        print(f"{self.name} says woof")


dog1=Dog("Buddy",3) # ovo je object
dog1.bark()  # use objest to call a bark methods
dog2=Dog("Lucy",4)
dog2.bark()

```
